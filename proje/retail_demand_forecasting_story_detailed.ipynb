```python
def dark_table(frame, formats=None, caption=None):
    styler = (
        frame.style
        .format(formats or {}, na_rep="—")
        .set_properties(
            **{
                "background-color": "#0f172a",
                "color": "#cbd5e1",
                "border-color": "#334155",
                "font-family": "Inter, Segoe UI, sans-serif",
            }
        )
        .set_table_styles(
            [
                {
                    "selector": "th",
                    "props": [
                        ("background", "linear-gradient(90deg,#312e81,#1e3a8a)"),
                        ("color", "#e0e7ff"),
                        ("border", "1px solid #475569"),
                        ("padding", "9px"),
                    ],
                },
                {
                    "selector": "td",
                    "props": [
                        ("border", "1px solid #334155"),
                        ("padding", "8px"),
                    ],
                },
                {
                    "selector": "caption",
                    "props": [
                        ("color", "#7dd3fc"),
                        ("font-weight", "800"),
                        ("padding", "8px"),
                    ],
                },
            ]
        )
    )
    return styler.set_caption(caption) if caption else styler


def dark_plot(fig, title):
    fig.update_layout(
        title=title,
        template="plotly_dark",
        paper_bgcolor="#0a0e1a",
        plot_bgcolor="#0a0e1a",
        font={"color": "#cbd5e1", "family": "Inter, Segoe UI, sans-serif"},
        title_font={"color": "#e0e7ff", "size": 18},
        legend_title_text="",
        margin={"l": 55, "r": 30, "t": 70, "b": 55},
    )
    fig.update_xaxes(gridcolor="#1e293b", zerolinecolor="#334155")
    fig.update_yaxes(gridcolor="#1e293b", zerolinecolor="#334155")
    return fig


display(pd.DataFrame({"Sunum yardımcısı": ["dark_table", "dark_plot"], "Durum": ["Hazır", "Hazır"]}))
```

### 05 · Gerçek test sonuçlarının iş hedefleriyle karşılaştırılması

Aşağıdaki kod hücresi yalnızca bu adımı gerçekleştirir.

```python
success_targets = {
    "KG": {"PR-AUC": (0.80, "higher"), "F1": (0.75, "higher"), "MAE": (2.00, "lower"), "WAPE": (0.90, "lower"), "|Bias|": (0.20, "lower")},
    "ADT": {"PR-AUC": (0.65, "higher"), "F1": (0.60, "higher"), "MAE": (1.25, "lower"), "WAPE": (1.00, "lower"), "|Bias|": (0.30, "lower")},
}

business_metric_rows = []
for unit in ("KG", "ADT"):
    test = model_metadata["test_metrics"][unit]
    actuals = {
        "PR-AUC": test["occurrence"]["pr_auc"],
        "F1": test["occurrence"]["f1"],
        "MAE": test["quantity_end_to_end"]["mae"],
        "WAPE": test["quantity_end_to_end"]["wape"],
        "|Bias|": abs(test["quantity_end_to_end"]["bias"]),
    }
    for metric, actual in actuals.items():
        target, direction = success_targets[unit][metric]
        achievement = (actual / target if direction == "higher" else target / max(actual, 1e-12)) * 100
        business_metric_rows.append(
            {
                "Birim": unit,
                "Metrik": metric,
                "Test sonucu": actual,
                "Hedef": target,
                "Hedefin % kaçı": achievement,
                "Durum": "HEDEFTE" if achievement >= 100 else "GELİŞTİRİLMELİ",
            }
        )

business_metrics = pd.DataFrame(business_metric_rows)
display(
    dark_table(
        business_metrics,
        {"Test sonucu": "{:.4f}", "Hedef": "{:.2f}", "Hedefin % kaçı": "{:.1f}%"},
        "Gerçek test sonuçları ve proje kabul hedefleri",
    ).map(
        lambda value: "color:#86efac;font-weight:800;" if value == "HEDEFTE" else "color:#fda4af;font-weight:800;",
        subset=["Durum"],
    )
)
```

### 59 · Permutation importance fonksiyonunun tanımlanması

Aşağıdaki kod hücresi yalnızca bu adımı gerçekleştirir.

```python
def permutation_rank(unit, stage, sample_size=1200):
    subset = interpretation_data.loc[interpretation_data["unit"].eq(unit)].copy()
    if stage == "quantity":
        subset = subset.loc[subset["target_demand"].gt(0)]
        target = "target_demand"
        scoring = "neg_mean_absolute_error"
    else:
        target = "demand_occurs"
        scoring = "average_precision"
    subset = subset.sample(min(sample_size, len(subset)), random_state=42)
    result = permutation_importance(
        deployed_models[unit][stage],
        subset[model_metadata["feature_columns"]].astype(np.float32),
        subset[target],
        n_repeats=2,
        random_state=42,
        scoring=scoring,
        n_jobs=1,
    )
    return (
        pd.DataFrame(
            {
                "Feature": model_metadata["feature_columns"],
                "Önem": result.importances_mean,
                "Önem_std": result.importances_std,
                "Birim": unit,
                "Aşama": stage,
            }
        )
        .sort_values("Önem", ascending=False)
        .reset_index(drop=True)
    )

display(pd.DataFrame({"Yorumlama yöntemi": ["Permutation importance"], "Tekrar": [2], "Örneklem/model": [1200]}))
```

### 72 · Validation threshold taramasının yapılması

Aşağıdaki kod hücresi yalnızca bu adımı gerçekleştirir.

```python
threshold_rows = []
for unit, y_true, probability in (
    ("KG", kg_validation["demand_occurs"].to_numpy(), kg_validation_probability),
    ("ADT", adt_validation["demand_occurs"].to_numpy(), adt_validation_probability),
):
    for threshold in np.linspace(0.05, 0.95, 91):
        decision = probability >= threshold
        threshold_rows.append(
            {
                "Birim": unit,
                "Threshold": threshold,
                "Precision": precision_score(y_true, decision, zero_division=0),
                "Recall": recall_score(y_true, decision, zero_division=0),
                "F1": f1_score(y_true, decision, zero_division=0),
            }
        )
threshold_curve = pd.DataFrame(threshold_rows)
display(pd.DataFrame({"Denenen eşik / birim": [91], "Toplam değerlendirme": [len(threshold_curve)]}))
```
